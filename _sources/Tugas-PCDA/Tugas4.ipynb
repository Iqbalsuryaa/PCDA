```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.datasets import fetch_olivetti_faces

# Menggunakan dataset citra wajah Yale; sebagai contoh, kita akan gunakan Olivetti Faces dari sklearn
data = fetch_olivetti_faces()
X = data.images
y = data.target

# Meresapkan citra dari matriks ke vektor
n_samples, h, w = X.shape
X = X.reshape(n_samples, h * w)

# Membagi dataset dengan rasio pelatihan 40% dan pengujian 60%
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.6, random_state=42)

# Mencetak hasil dengan berbagai dimensi PCA
print("Hasil klasifikasi menggunakan berbagai dimensi PCA:")
for n_components in range(10, 51, 10):
    # Menggunakan PCA untuk reduksi dimensi
    pca = PCA(n_components=n_components, whiten=True, random_state=42)
    X_train_pca = pca.fit_transform(X_train)
    X_test_pca = pca.transform(X_test)

    # Melatih model SVM pada data latih
    svm = SVC(kernel='linear', class_weight='balanced', random_state=42)
    svm.fit(X_train_pca, y_train)

    # Prediksi pada data uji
    y_pred = svm.predict(X_test_pca)

    # Menghitung akurasi
    accuracy = accuracy_score(y_test, y_pred)
    print(f"Dimensi PCA: {n_components}, Akurasi: {accuracy:.2f}")

print("\nSelesai menjalankan pelatihan dan pengujian.")
```
