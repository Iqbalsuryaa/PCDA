**Operator Gradien**

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load the image in grayscale
image = cv2.imread('/content/kucing citra.jpeg', cv2.IMREAD_GRAYSCALE)

# Gradient in the x-direction
gradient_x = cv2.Sobel(image, cv2.CV_64F, 1, 0, ksize=3)  # Sobel derivative in X direction

# Gradient in the y-direction
gradient_y = cv2.Sobel(image, cv2.CV_64F, 0, 1, ksize=3)  # Sobel derivative in Y direction

# Magnitude of the gradient
gradient_magnitude = cv2.magnitude(gradient_x, gradient_y)

# Convert gradients to 8-bit for visualization
gradient_x = cv2.convertScaleAbs(gradient_x)
gradient_y = cv2.convertScaleAbs(gradient_y)
gradient_magnitude = cv2.convertScaleAbs(gradient_magnitude)

# Plot the results
plt.figure(figsize=(12, 6))

plt.subplot(1, 4, 1)
plt.imshow(image, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 4, 2)
plt.imshow(gradient_x, cmap='gray')
plt.title('Gradient X')
plt.axis('off')

plt.subplot(1, 4, 3)
plt.imshow(gradient_y, cmap='gray')
plt.title('Gradient Y')
plt.axis('off')

plt.subplot(1, 4, 4)
plt.imshow(gradient_magnitude, cmap='gray')
plt.title('Gradient Magnitude')
plt.axis('off')

plt.tight_layout()
plt.show()
```

**Operator Robert's Cros**

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load the image in grayscale
image = cv2.imread('/content/kucing citra.jpeg', cv2.IMREAD_GRAYSCALE)

# Define the Robert's Cross kernels
kernel_x = np.array([[1, 0], [0, -1]], dtype=float)  # Kernel for detecting edges in the x-direction
kernel_y = np.array([[0, 1], [-1, 0]], dtype=float)  # Kernel for detecting edges in the y-direction

# Apply convolution using the kernels
roberts_x = cv2.filter2D(image, cv2.CV_64F, kernel_x)
roberts_y = cv2.filter2D(image, cv2.CV_64F, kernel_y)

# Calculate the magnitude of the gradient
roberts_magnitude = cv2.magnitude(roberts_x, roberts_y)

# Convert to 8-bit for visualization
roberts_magnitude_8bit = cv2.convertScaleAbs(roberts_magnitude)

# Display the results
plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.imshow(image, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(roberts_magnitude_8bit, cmap='gray')
plt.title("Robert's Cross Edge Detection")
plt.axis('off')

plt.tight_layout()
plt.show()
```

**Operator Prewit**

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load the image in grayscale
image = cv2.imread('/content/kucing citra.jpeg', cv2.IMREAD_GRAYSCALE)

# Define the Prewitt kernels
kernel_x = np.array([[1, 0, -1],
                      [1, 0, -1],
                      [1, 0, -1]], dtype=float)  # Kernel for detecting edges in the x-direction

kernel_y = np.array([[1, 1, 1],
                      [0, 0, 0],
                      [-1, -1, -1]], dtype=float)  # Kernel for detecting edges in the y-direction

# Apply convolution using the kernels
prewitt_x = cv2.filter2D(image, cv2.CV_64F, kernel_x)
prewitt_y = cv2.filter2D(image, cv2.CV_64F, kernel_y)

# Calculate the magnitude of the gradient
prewitt_magnitude = cv2.magnitude(prewitt_x, prewitt_y)

# Convert to 8-bit for visualization
prewitt_magnitude_8bit = cv2.convertScaleAbs(prewitt_magnitude)

# Display the results
plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.imshow(image, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(prewitt_magnitude_8bit, cmap='gray')
plt.title('Prewitt Edge Detection')
plt.axis('off')

plt.tight_layout()
plt.show()
```
